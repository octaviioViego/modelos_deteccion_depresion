```python
## Importar dependencias
import pandas as pd
import glob
import os
from pathlib import Path
from dotenv import load_dotenv


# Cargar variables del archivo .env
load_dotenv()

## Diccionario de PATH´s
PATHS = {
    "Terea2_train_2023": Path(os.environ["path_metalriskES2023_task2_train"]),
    "Terea2_trial_2023": Path(os.environ["path_metalriskES2023_task2_trial"]),
    "Terea2_test_2023": Path(os.environ["path_metalriskES2023_task2_test"]),
}

## Obtenemos las rutas del PATH en variables
datos_tarea2_train_2023 =  PATHS["Terea2_train_2023"]
datos_tarea2_test_2023 =   PATHS["Terea2_test_2023"] 
datos_tarea2_trial_2023 =  PATHS["Terea2_trial_2023"]
```

```python
## Recuperacion y manipulacion de los archivos JSON del Train
lista_mensajes_train = []
lista_nombres_archivos_train: list[str] = []
cantidad_archivos: int = 0

for archivo in datos_tarea2_train_2023.glob("*.json"):
    pd_train_archivo = pd.read_json(archivo)
    lista_nombres_archivos_train.append(str(archivo))
    lista_mensajes_train.append(pd_train_archivo)
    cantidad_archivos += 1
```
