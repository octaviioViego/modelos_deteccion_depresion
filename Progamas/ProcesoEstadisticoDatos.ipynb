```python
## Importar dependencias
import glob
import os
from io import StringIO
from pathlib import Path
from dotenv import load_dotenv
import pandas as pd
```

```python
## Recuperacion de los archivos CSV

# Cargar variables del archivo .env
load_dotenv()

## Diccionario de PATHs
PATHS = {
    "Tarea2_CSV": Path(os.environ["path_metalriskES2023_task2_CSV"]),
}

## Variables de rutas
# Tarea 2 rutas
ruta_guardado_tarea2_csv = PATHS["Tarea2_CSV"]

# Nombres de archivos 
nombre_archivo_tarea2_A_csv = "Tarea_A.csv"
nombre_archivo_tarea2_B_csv = "Tarea_B.csv"
nombre_archivo_tarea2_C_csv = "Tarea_C.csv"
nombre_archivo_tarea2_D_csv = "Tarea_D.csv"


 ## Cargamos las etiquetas GOLD de la tarea 2 A
ruta_tarea2_a_csv= os.path.join(ruta_guardado_tarea2_csv, 
                                nombre_archivo_tarea2_A_csv)


datos_CSV_tarea2 = pd.read_csv(ruta_tarea2_a_csv)


```

```python
#print(datos_CSV_tarea2["sujeto"])
#print(type(datos_CSV_tarea2["mensaje"]))
#print(datos_CSV_tarea2["fecha"])
#print(datos_CSV_tarea2["etiqueta_gold"])

lista_mensajes = datos_CSV_tarea2["mensaje"].tolist()


mas_largo = max(lista_mensajes, key=len)
mar_corto = min(lista_mensajes, key=len)

media = sum(len(texto) for texto in lista_mensajes) / len(lista_mensajes)
print(f"Más cargo: {len(mas_largo)}")
print(f"Más corto: {mar_corto}")
print(f"Media: {media}")


```
