## Manipulacion de los datos del TEST
- Se manipularion los datos para tener archivoa CSV con:
    - Nombre de sujeto
    - Mensaje
    - Fecha
    - Ronda

Esto para tener una mejor manipulación y entendimiento de los datos en las pruebas

```python
## Importar dependencias
import glob
import os
from pathlib import Path
from dotenv import load_dotenv
import pandas as pd
```

```python
# Cargar variables del archivo .env
load_dotenv()

## Diccionario de PATHs
PATHS = {
    # Tarea 2 
    "Terea2_test_2023": Path(os.environ["path_metalriskES2023_task2_test"]),
    "Terea2_test_2023_gold": Path(os.environ["path_metalriskES2023_task2_test_gold"]),
    "Tarea2_CSV": Path(os.environ["path_metalriskES2023_task2_CSV_test"]),
}

## Variables de rutas
# Tarea 2 rutas
datos_tarea2_test_2023 = PATHS["Terea2_test_2023"]
datos_tarea2_A_test_2023_gold = PATHS["Terea2_test_2023_gold"]
ruta_guardado_tarea2_csv = PATHS["Tarea2_CSV"]

# Nombres de archivos 
nombre_etiquetas_gold_A_2023 = "gold_a.txt"
nombre_archivo_tarea2_A_csv = "Tarea_A.csv"

nombre_etiquetas_gold_B_2023 = "gold_b.txt"
nombre_archivo_tarea2_B_csv = "Tarea_B.csv"

nombre_etiquetas_gold_C_2023 = "gold_c.txt"
nombre_archivo_tarea2_C_csv = "Tarea_C.csv"

nombre_etiquetas_gold_D_2023 = "gold_d.txt"
nombre_archivo_tarea2_D_csv = "Tarea_D.csv"

```

```python
## Manipulacion y reorganizacion de los datos test de la Tarea 2 A , B , C.

# Los preparamos en una lista 
datos_a_manipular = [[nombre_etiquetas_gold_A_2023,nombre_archivo_tarea2_A_csv],
                     [nombre_etiquetas_gold_B_2023,nombre_archivo_tarea2_B_csv],
                     [nombre_etiquetas_gold_C_2023,nombre_archivo_tarea2_C_csv],
                     [nombre_etiquetas_gold_D_2023,nombre_archivo_tarea2_D_csv]]

for etiquetas_gold, nombre_archivos_csv in datos_a_manipular:
        
    ## Cargamos las etiquetas GOLD de la tarea 2 A
    etiquetas_gold_tarea2_A_2023 = os.path.join(datos_tarea2_A_test_2023_gold, 
                                                etiquetas_gold)

    # Leemos el archivo con etiquetas
    df_gold = pd.read_csv(etiquetas_gold_tarea2_A_2023, header=None, sep="\t", names=["raw"])

    #  Creamos un diccionario con las etiquetas
    diccionario_etiquetas = {}

    # Evaluamos el caso de las etiquetas GOLD Tarea 2 C. el cual tiene mas de una etiqueta.
    if(etiquetas_gold !=nombre_etiquetas_gold_D_2023):
        for item in df_gold["raw"]:
            if "," in str(item):
                sujeto, etiqueta = str(item).split(",")
                diccionario_etiquetas[sujeto.strip()] = etiqueta.strip()

    else:
        for item in df_gold["raw"]:
            if "," in str(item):
                sujeto, etiqueta_afavor, etiqueta_contra, etiqueta_otro, control  = str(item).split(",")
                diccionario_etiquetas[sujeto.strip()] = [etiqueta_afavor.strip(),
                                                         etiqueta_contra.strip(),
                                                         etiqueta_otro.strip(),
                                                         control.strip()]
        

    ## Leer archivos JSON y aplanar filas
    datos_usuario_reorganizado = []

    for archivo in datos_tarea2_test_2023.glob("*.json"):

        #sujeto_nombre = archivo
        #etiqueta = diccionario_etiquetas.get(sujeto_nombre, None)
        
        df_json = pd.read_json(archivo)
        
        # Iterar cada mensaje individual dentro del archivo del sujeto (Guardamos el nombre,mensaje,fecha, etiqueta).
        for _, fila in df_json.iterrows():
            etiqueta = diccionario_etiquetas.get(fila["nick"], None)
            datos_usuario_reorganizado.append([ fila["nick"],
                                                fila["round"],   
                                                fila["message"],
                                                fila["date"],
                                                etiqueta])

    ## Convertir a DataFrame final
    df_resultado = pd.DataFrame(datos_usuario_reorganizado,
                                columns=["sujeto","round/asalto", "mensaje", "fecha", "etiqueta_gold"],)

    ## Guardamos en archivo CSV
    etiquetas_gold_tarea2_A_test_2023 = os.path.join(ruta_guardado_tarea2_csv, 
                                                nombre_archivos_csv)

    print(etiquetas_gold_tarea2_A_test_2023)
    df_resultado.to_csv(etiquetas_gold_tarea2_A_test_2023, index=False)
```
